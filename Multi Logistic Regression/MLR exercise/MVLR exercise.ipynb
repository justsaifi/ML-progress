```python
import pandas as pd
df=pd.read_csv("newdata.csv")
df
```

```python
df['test_score(out of 10)'].median()
```

```python
import math
df.median_test=math.floor(df['test_score(out of 10)'].median())
```

```python
df.median_test
```

```python
df['test_score(out of 10)']=df['test_score(out of 10)'].fillna(df.median_test)
```

```python
df
```

```python
df['experience']=df['experience'].astype('string')
```

```python
from word2number import w2n

df["experience"] = df["experience"].apply(w2n.word_to_num)
```

```python
df['test_score(out of 10)']=df['test_score(out of 10)'].astype(int)
```

```python
print(df.dtypes)
```

```python
df
```

```python
from sklearn import linear_model
reg=linear_model.LinearRegression()
```

```python
reg.fit(df[['experience','test_score(out of 10)','interview_score(out of 10)']],df[['salary(S)']])
```

```python
reg.predict([[2,9,6]])
```

```python
reg.predict([[12,10,10]])
```
