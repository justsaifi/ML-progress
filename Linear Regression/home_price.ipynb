```python
import pandas as pd
```

```python
df=pd.read_csv("home prices.csv")
df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df.Area,df.price)
plt.xlabel('Area')
plt.ylabel('price')
plt.title('Home prices')
plt.show()
```

```python
from sklearn import linear_model
```

```python
reg=linear_model.LinearRegression()
```

```python
reg.fit(df[['Area']],df[['price']])
```

```python
reg.predict([[3000]])
```

```python
reg.coef_
```

```python
reg.intercept_
```
