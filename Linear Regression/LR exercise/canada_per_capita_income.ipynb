```python
import pandas as pd
df=pd.read_csv("canada_per_capita_income.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df[['year']],df[['per capita income (US$)']])
plt.xlabel('year')
plt.ylabel('income')
```

```python
from sklearn import linear_model
```

```python
clf=linear_model.LinearRegression()
```

```python
clf.fit(df[['year']],df[['per capita income (US$)']])
```

```python
clf.predict([[2020]])
```

```python
import pandas as pd
d=pd.read_csv("year.csv")
d.head()
```

```python
I = clf.predict(d)
```

```python
d['income']= I
d
```

```python
d.to_csv("income.csv",index=False)
```
